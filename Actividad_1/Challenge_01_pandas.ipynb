## __Parte 4 - Estadisticos__  

Calcular:
> - media  
> - mediana  
> - mínimo  
> - máximo   
> - suma  
> - conteo   


```python
#Obteener la Media del total_gastado 
print("La media del total gastado es:", df['Total_Gastado'].mean())
```

```python
#Obteener la Mediana del total_gastado 
print("La mediana del total gastado es:", df['Total_Gastado'].median())
```

```python
#Obtener el mínimo del total gastado y máximo
print("El mínimo del total gastado es:", df['Total_Gastado'].min())
print("El máximo del total gastado es:", df['Total_Gastado'].max())
```

```python
#La suma del total gastado  
print("La suma del total gastado es:", df['Total_Gastado'].sum())
```

```python
#El conteo sobre la columna de ciudad
conteo_ciudad = df['Ciudad'].value_counts()
print("Conteo de observaciones por ciudad:\n", conteo_ciudad)
```
