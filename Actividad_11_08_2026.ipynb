Se realizará un agrupado por región para coonocer cuantas regiones tenemos

```python
agrupado_region = df.groupby("Region").agg({"Unidades_Vendidas": "sum"})
agrupado_region
```

Al ejecutarlo encontramos que nuestras regiones viene separadas, debido a la modificacion por máyusculas y minúsculas, por lo que haremos la modificación para que python realicé una mejor agrupación

```python
df["Region"] = df["Region"].str.strip()  # Elimina espacios al inicio y al final
df["Region"] = df["Region"].str.title()  #Convierte la primera letra de cada palabra a mayúscula y el resto a minúscula
agrupado_region = df.groupby("Region", as_index=True)["Unidades_Vendidas"].sum()
agrupado_region
```

```python
#Agrupado por categoria para conocer que tipo de categorias tenemos
agrupado_categoria = df.groupby("Categoria", as_index=True)["Unidades_Vendidas"].sum()
agrupado_categoria
```

```python
#Ahora como existen valores nulos, vamos analizar si vienen por alguna region en especifico
agrupado_nulos = df.groupby("Region").apply(lambda x: x.isnull().sum())
agrupado_nulos
```

```python
#Ahora como existen valores nulos, vamos analizar si vienen por alguna categoria en especifico
agrupado_categoria_nulos = df.groupby("Categoria").apply(lambda x: x.isnull().sum())
agrupado_categoria_nulos
```
