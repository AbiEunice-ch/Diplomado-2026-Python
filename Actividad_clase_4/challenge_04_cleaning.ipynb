duplicado, ya que algunos registros presentan diferencias en otras variables,
como `Ventas`.

Por esta razón, únicamente se considerarán duplicados aquellos registros

```python
# Vuelve a partir del dataset original para no conservar conversiones fallidas
df_limpio = df_corruptas.drop_duplicates().copy()

df_limpio["Precio"] = pd.to_numeric(
    df_limpio["Precio"]
        .astype("string")
        .str.strip()
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
        .replace({"sin dato": pd.NA, "": pd.NA}),
    errors="coerce"
)

df_limpio[["Precio"]].isna().sum()
```

```python
# Sustituir nombres de meses en español antes de interpretar las fechas
meses = {
    "Enero": "January", "Febrero": "February", "Marzo": "March",
    "Abril": "April", "Mayo": "May", "Junio": "June",
    "Julio": "July", "Agosto": "August", "Septiembre": "September",
    "Octubre": "October", "Noviembre": "November", "Diciembre": "December"
}

fechas = df_limpio["Fecha"].astype("string").str.strip()
for espanol, ingles in meses.items():
    fechas = fechas.str.replace(espanol, ingles, case=False, regex=False)

# mixed admite los distintos formatos presentes en el archivo.
# Revisa las fechas ambiguas (por ejemplo, 05-03-2025) antes de fijar
# si el archivo las expresa como día-mes o mes-día.
df_limpio["Fecha"] = pd.to_datetime(fechas, format="mixed", errors="coerce")
```

```python
#Revu¿isaremos las variables de Producto y Region
print("PRODUCTO")
print(df_limpio["Producto"].value_counts(dropna=False).to_string())

print("\nREGION")
print(df_limpio["Region"].value_counts(dropna=False).to_string())
```

Se observa que los datos vienen sucios en producto y region, por lo que como siguiente paso normalizaremos estas variables

```python
# Con este código se eliminan espacios en blanco, se pasa a minúsculas y luego a mayúscula inicial
df_limpio["Producto"] = (
    df_limpio["Producto"]
    .astype("string")
    .str.strip()
    .str.lower()
    .str.title()
)

df_limpio["Region"] = (
    df_limpio["Region"]
    .astype("string")
    .str.strip()
    .str.lower()
    .str.title()
)

print(df_limpio["Producto"].value_counts(dropna=False))
print(df_limpio["Region"].value_counts(dropna=False))
```

### Ahora revisaremos faltantes antes de corregir

```python
resumen_faltantes = pd.DataFrame({
    "faltantes": df_limpio.isna().sum(),
    "porcentaje": (df_limpio.isna().mean() * 100).round(2)
})

resumen_faltantes
```

```python
#Se necesita convertir a numérico para poder hacer análisis estadísticos y visualizaciones. Se eliminan los símbolos de moneda y comas, y se reemplazan los valores "sin dato" y vacíos por NaN.
df_limpio["Precio"] = pd.to_numeric(
    df_limpio["Precio"]
    .astype("string")
    .str.strip()
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .replace({"sin dato": pd.NA, "": pd.NA}),
    errors="coerce"
)

df_limpio["Edad"] = pd.to_numeric(df_limpio["Edad"], errors="coerce")
df_limpio["Ventas"] = pd.to_numeric(df_limpio["Ventas"], errors="coerce")

print(df_limpio[["Edad", "Precio", "Ventas"]].dtypes)
```

```python
# Convertir a nulos las edades negativas y la edad 397
df_limpio["Edad"] = pd.to_numeric(df_limpio["Edad"], errors="coerce")

edades_invalidas = (df_limpio["Edad"] < 0) | (df_limpio["Edad"] == 397)

print("Edades inválidas encontradas:", edades_invalidas.sum())

df_limpio.loc[edades_invalidas, "Edad"] = pd.NA

print("Nulos en Edad después del cambio:", df_limpio["Edad"].isna().sum())
```

```python
mediana_edad = df_limpio["Edad"].median()
print("Mediana utilizada:", mediana_edad)

df_limpio["Edad"] = df_limpio["Edad"].fillna(mediana_edad)
```

```python
resumen_nulos = pd.DataFrame({
    "Nulos": df_limpio.isna().sum(),
    "Porcentaje": (df_limpio.isna().mean() * 100).round(2)
})

resumen_nulos
```

```python
# Uniformar espacios y mayúsculas
df_limpio["Region"] = (
    df_limpio["Region"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Unir las distintas formas de indicar una región desconocida
df_limpio["Region"] = df_limpio["Region"].replace({
    "?": "Sin Especificar",
    "Sin Informacion": "Sin especificar"
})

df_limpio["Region"] = df_limpio["Region"].fillna("Sin Especificar")

df_limpio["Region"].value_counts(dropna=False)
```

```python
#Normalizemos Precio
print("Precios nulos:", df_limpio["Precio"].isna().sum())

df_limpio.loc[
    df_limpio["Precio"].isna(),
    ["Producto", "Precio", "Ventas"]
].head(10)
```

```python
mediana_por_producto = df_limpio.groupby("Producto")["Precio"].transform("median")

df_limpio["Precio"] = df_limpio["Precio"].fillna(mediana_por_producto)

print("Precios que siguen nulos:", df_limpio["Precio"].isna().sum())
```

```python
df_limpio.loc[
    df_limpio["Precio"].isna(),
    ["Producto", "Precio", "Ventas"]
]
```

```python
df_limpio.isna().sum()
```

Los precios faltantes se imputaron con la mediana de los precios válidos del mismo producto. Esta decisión conserva las observaciones y toma como referencia productos comparables; los valores imputados son estimaciones, no precios observados.

```python
indices_fecha_nula = df_limpio.index[df_limpio["Fecha"].isna()]

revision_fechas = pd.DataFrame({
    "Fecha_original": df_corruptas.loc[indices_fecha_nula, "Fecha"],
    "Fecha_convertida": df_limpio.loc[indices_fecha_nula, "Fecha"]
})

print("Originalmente vacías:",
      revision_fechas["Fecha_original"].isna().sum())

print("Tenían texto, pero no se convirtieron:",
      revision_fechas["Fecha_original"].notna().sum())

print("\nEjemplos de fechas que no se convirtieron:")
print(
    revision_fechas.loc[
        revision_fechas["Fecha_original"].notna(),
        "Fecha_original"
    ].value_counts().head(20)
)
```

```python
print("Fechas originalmente vacías:",
      revision_fechas["Fecha_original"].isna().sum())

print("Fechas con valor imposible 32/13/2026:",
      revision_fechas["Fecha_original"].eq("32/13/2026").sum())

print("Total de fechas nulas en el dataset limpio:",
      df_limpio["Fecha"].isna().sum())
```

Después de eliminar los duplicados, se identificaron 309 fechas originalmente vacías y 286 registros con el valor imposible 32/13/2026. Estos últimos se transformaron en valores nulos durante la conversión a fecha. Se conservaron los 595 valores como NaT, ya que no existe información suficiente para recuperar las fechas reales sin inventarlas. Los registros se mantienen porque sus demás variables todavía pueden utilizarse en análisis que no dependan de la fecha

```python
#Ahora revisaremos los valores de Precio y Ventas para ver si hay valores negativos o cero, que no deberían existir en este contexto.
print(df_limpio[["Precio", "Ventas"]].describe())

print("\nPrecios menores o iguales a cero:")
print(df_limpio.loc[df_limpio["Precio"] <= 0,
                    ["ID", "Producto", "Precio", "Ventas"]].head(20))
print("Total:", (df_limpio["Precio"] <= 0).sum())

print("\nVentas menores o iguales a cero:")
print(df_limpio.loc[df_limpio["Ventas"] <= 0,
                    ["ID", "Producto", "Precio", "Ventas"]].head(20))
print("Total:", (df_limpio["Ventas"] <= 0).sum())
```

```python
for columna in ["Precio", "Ventas"]:
    q1 = df_limpio[columna].quantile(0.25)
    q3 = df_limpio[columna].quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    extremos = df_limpio[
        (df_limpio[columna] < limite_inferior) |
        (df_limpio[columna] > limite_superior)
    ]

    print(f"\n{columna}")
    print(f"Límites IQR: {limite_inferior:.2f} a {limite_superior:.2f}")
    print(f"Valores señalados: {len(extremos)}")
    print(extremos[["ID", "Producto", "Precio", "Ventas"]].head(10))
```

```python
precios_negativos = df_limpio["Precio"] < 0
precios_cero = df_limpio["Precio"] == 0

print("Precios negativos:", precios_negativos.sum())
print("Precios iguales a cero:", precios_cero.sum())

# Guardar una muestra para documentar qué se corrigió
df_limpio.loc[
    precios_negativos | precios_cero,
    ["ID", "Producto", "Precio", "Ventas"]
].head(10)
```

```python
# Los precios no positivos se consideran inválidos
df_limpio.loc[df_limpio["Precio"] <= 0, "Precio"] = float("nan")

# Imputar usando únicamente precios válidos del mismo producto
mediana_por_producto = df_limpio.groupby("Producto")["Precio"].transform("median")
df_limpio["Precio"] = df_limpio["Precio"].fillna(mediana_por_producto)

print("Precios nulos restantes:", df_limpio["Precio"].isna().sum())
print("Precios no positivos restantes:", (df_limpio["Precio"] <= 0).sum())
```

Se detectaron precios negativos, que no son válidos como precios de productos. Se trataron como faltantes y se imputaron con la mediana de precios positivos del mismo producto. Los valores extremos de ventas se señalaron para revisión, sin eliminarlos automáticamente, ya que un importe inusual puede corresponder a una venta real.

```python
print("Ventas más bajas:")
display(
    df_limpio.nsmallest(10, "Ventas")[
        ["ID", "Producto", "Region", "Precio", "Ventas"]
    ]
)

print("Ventas más altas:")
display(
    df_limpio.nlargest(10, "Ventas")[
        ["ID", "Producto", "Region", "Precio", "Ventas"]
    ]
)

print("Ventas menores o iguales a cero:",
      (df_limpio["Ventas"] <= 0).sum())
```

Se identificaron valores extremos en Ventas mediante el rango intercuartílico. Se revisaron los importes más bajos y altos. Los valores inusuales se conservaron cuando no había evidencia de un error, porque eliminarlos modificaría la distribución y la media de ventas.
